Ejercicios de Variables y Tipos de Datos Basicos

1. Tipos de Datos

```python
#Definicion de un nro entero
edad = 20
print("Entero:", edad, "| Tipo:", type(edad))

#Definicion de un nro decimal
altura = 1.75
print("Decimal:", altura, "| Tipo:", type(altura))

#Definicion de un string
nombre = "Pedro Damian"
print("String:", nombre, "| Tipo:", type(nombre))

#Definicion de un booleano
es_mayor = True
print("Boolean:", es_mayor, "| Tipo:", type(es_mayor))
```

2. Operaciones con Strings

```python
#Nombre de la persona
nombre = "Damian"

#Apellido de la persona
apellido = "Rojas"

print("Concatenación simple:", nombre + apellido)
print("Concatenación con punto:", nombre + apellido + ".")
print("Repetición x3:", nombre * 3)
```

3. Slicing

```python
nombreCompleto = "Damian Rojas"

#Slicing [0:6]: inicia en el índice 0 (D) y termina antes del índice 6 (espacio)
print("Solo el nombre de la persona:", nombreCompleto[0:6])
```

Ejercicios de Operaciones y Expresiones

1. Media Aritmetica

```python
#Notas del estudiante Damian y la media de sus notas
nota1 = 8.5
nota2 = 9.0
nota3 = 7.5
nota4 = 10.0
nota5 = 6.0
media = (nota1 + nota2 + nota3 + nota4 + nota5) / 5
print("Media de las notas de Damian:", media)
```

2. Redondeo

```python
operacion = (365 / 12) * 14.7

print("El resultado de la operación es:", operacion)

#Usando la función round() para redondear el resultado a 2 decimales
print("El resultado de la operación redondeada a 2 decimales es:", round(operacion, 2))
```

3. Operadores Logicos

```python
#Credenciales de usuario
username = "DamianRojas"
password = "Tokio"

#Verificamos si el username tiene entre 3 y 10 caracteres y si el password es "Tokio" o "Python"
comprobacion_user = len(username) >= 3 and len(username) <= 10
comprobacion_pass = password == "Tokio" or password == "Python"

#Verificamos si username y password son verdaderos (True) o falsos (False)
print("Username es:", comprobacion_user)
print("Password es:", comprobacion_pass)
```

Operadores I/O

1. Operadores de Asignacion

```python
num1 = 10
num2 = 5

#Aumenta su valor en 1
num1 += 1
print("Resultado con +=:", num1)
#Disminuye su valor en 2
num2 -= 2
print("Resultado con -=:", num2)
#Multiplica su valor por 3
num1 *= 3
print("Resultado con *=:", num1)
#Divide su valor entre 2
num2 /= 2
print("Resultado con /=:", num2)

```

2. Funcion input()

```python
#Ejtrada de datos por teclado sin convertirlos a otro tipo de dato
nro_inicial = input("Ingrese un número: ")
print("El nro ingresado es:", nro_inicial, "| Tipo:", type(nro_inicial))

#Entrada de datos por teclado convirtiéndolos a int
nro_entero = int(input("Ingrese un número entero: "))
print("El nro ingresado es:", nro_entero, "| Tipo:", type(nro_entero))

#Entrada de datos por teclado convirtiéndolos a float
nro_decimal = float(input("Ingrese un número decimal: "))
print("El nro ingresado es:", nro_decimal, "| Tipo:", type(nro_decimal))
```

3. Formateo de Numeros

```python
num_int = int(input("Ingrese un número entero: "))
num_float = float(input("Ingrese un número decimal: "))

#Formateo de números enteros y decimales
print(f"El nro entero formateado queda como: {num_int:05d}")
print(f"El nro decimal formateado queda como: {num_float:09.3f}")
```

4. Formas de formatear

```python
altura = float(input("Ingrese su altura: "))
peso = float(input("Ingrese su peso: "))

# 1. Formateo con f-strings
print(f"La altura es de {altura:.2f} metros y el peso es de {peso:.3f} KG")

# 2. Formateo con el método .format() posicional
print("La altura es de {:.2f} metros y el peso es de {:.3f} KG".format(altura, peso))

# 3. Formateo con el método .format() especificando el orden de los argumentos
print("La altura es de {0:.2f} metros y el peso es de {1:.3f} KG".format(altura, peso))
```
