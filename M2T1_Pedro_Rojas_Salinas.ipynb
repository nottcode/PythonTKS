 Manipulacion Avanzada de Texto

1. Longitud de una cadena

```python
mensaje = "Hola, bienvenidos al curso de Python"

#Funcion para obtener el nro total de caracteres de la cadena
caracteres_mensaje = len(mensaje)

print("El nro total de caracteres de la cadena es: ", caracteres_mensaje)
```

2. Indices de cadenas

```python
texto = "Python"

#Extraemos la primera letra usando indices
print("La primera letra es: ", texto[0])

#Extraemos la ultima letra
print("La ultima letra es: ", texto[-1])
```

3. Extraer partes de una direccion

```python
direccion = "Calle Falsa, 123, Springfield"

#Extraemos el nombre de la calle
calle = direccion[0:11]
print("El nombre de la calle es: ", calle)

#Extraemos el nro
numero = direccion[13:16]
print("El nro de la calle es: ", numero)

#Extraemos ahora la ciudad
ciudad = direccion[18:]
print("El nombre de la ciudad es: ", ciudad)
```

4.  Crear un identificador unico con slicing y concatenacion

```python
nombre = "Damian"
apellido = "Rojas"
año_nacimiento = "2006"

#Extramos las primeras 3 letras del nombre, apellido y las 2 ultimas del año de nacimiento
letras_nombre = nombre[:3]
letras_apellido = apellido[:3]
dig_años = año_nacimiento[-2:]

#Ahora concatenamos para poder mostrar el texto DamRoj06
texto_final = letras_nombre + letras_apellido + dig_años

print(texto_final)
```

5. Modificar una cadena usando slicing

```python
palabra = "robot"

#palabra[0] = "R" esto nos lanza un TypeError

#Creamos una nueva palabra usando slicing y concatenacion
palabra_modificada = "R" + palabra[1:]

print("La palabra modificada es: ", palabra_modificada)
```

6.  Analizador de palabras con split, count y upper/lower

```python
texto = "Python es POTENTE. Aprender python es útil. PYTHON abre puertas."
objetivo = "Python"

#Creamos una version en minusculas del texto para una mejor busqueda
texto_lower = texto.lower()
objetivo_lower = objetivo.lower()

#Con el metodo .split contamos cuantas veces aparece el objetivo
lista_palabras = texto_lower.split()
conteo_palabras = lista_palabras.count(objetivo_lower)

#Y con .count para saber cuantas veces aparece la subcadena
aparicion_subcadena = texto_lower.count(objetivo_lower)

#Como extra reconstruimos el texto original pero con el objetico en mayusculas
objetivo_upper = objetivo.upper()
texto_resaltado = (objetivo_upper + texto[6:28] + objetivo_upper + texto[34:44] + objetivo_upper + texto[50:])

print("Total (subcadenas): ", aparicion_subcadena)
print("Total (palabra exacta): ", conteo_palabras)
print("Texto resaltado: ", texto_resaltado)
```
