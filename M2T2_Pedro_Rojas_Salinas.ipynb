1. Orden de los números

- Realiza un programa que lea tres números por teclado y
evalúe lo siguiente:
o Si los números están en orden ascendente.
o Si los números no están en orden ascendente.
o Si hay un error porque el primer número introducido es 0.
Nota: No uses listas en este ejercicio.

```python
#Ingresamos 3 nros por teclado
nro1 = float(input("Ingrese un nro: "))
nro2 = float(input("Ingrese otro nro: "))
nro3 = float(input("Ingrese el ultimo nro: "))

#Creamos las validaciones correspondientes
if nro1 == 0: #vemos si el primer nro es 0
    print("Error: el primer nro introducido es 0")
elif nro1 < nro2 < nro3: #aqui podemos saber si el orden de nros introducimdos es ascendente
    print("Los nros estan en orden ascendente")
else: #aqui si no tienen dicho orden
    print("Los nros no estan en orden ascendente")
```

3. Sumatorio de los primeros N números

- Realiza un programa que realice el
sumatorio de los primeros N números, empezando en 0, y siendo N el número límite
que el usuario proporciona por teclado.
Realiza un programa que realice el
sumatorio de los primeros N números, empezando en 0, y siendo N el número límite
que el usuario proporciona por teclado.

```python
#pedimos el nro que sera N
limite_n = int(input("Ingrese un numero limite: "))

#enm esta variable vamos a guardar la suma total
sumatoria = 0

#este bucle recorre de 0 a N-1
for i in range(limite_n):
    sumatoria += i

#imprimimos le resultado final de la suma
print(f"El sumatorio de los primeros {limite_n} números (de 0 a {limite_n - 1}) es: {sumatoria}")
```

4. Contador de letras ‘a’

- Realiza un programa que lea letras una por una y cuente con
una variable cuantas veces se introduce la letra ‘a’

```python
contador = 0 #definimos un contador para las letras a encontradas 
while True:
    letra = input("Introduzca una leta (para salir presione '.'): ")

    if letra == ".": #condicion para detener el bucle
        break

    if letra == "a":
        contador += 1   #cuando encuentra una letra a se suma 1 a esa variable

print("La cantidad de letras 'a' presionadas son de: ", contador)
```

5. Contador de letras ‘a’ en palabras

- Realiza un programa similar al anterior, pero que
lea letras completas y cuente cuantas veces aparece la letra ‘a’

```python
contador = 0

while True:
    palabra = input("Introduzca una palabra (para salir presione '.'): ")

    if palabra == ".": #condicion para detener el bucle
        break

    contador += palabra.count("a")   #cuando encuentra una letra a en esa palabra se suma 1 a esa variable

print("La cantidad de letras 'a' presionadas en cada palabra son de: ", contador)
```

6. Sumatorio de los números pares decrecientes

```python
#variable para la sumetoria
sumatorio_pares = 0

#buvle for que va de 1000 a 72 de 2 en 2 para agarrar nada mas los nros pares 
for numero in range(1000, 72, -2):
    sumatorio_pares += numero

#mostramos el resultado final
print(f"El sumatorio de los números pares entre 1000 y 73 es: {sumatorio_pares}")
```

7. Lista de números: análisis y operaciones

- Realiza un programa que solicite al usuario
un número que represente cuantos valores va a introducir. Posteriormente, lee todos
esos números, guárdalos en una lista y realiza las siguientes operaciones:
o La ordenación de la lisa de menor a mayor.
o El cálculo promedio.
o Identifica el valor máximo y mínimo, y la posición en la que se encuentran.

```python
#le pedimops al usuario cuantos numeros va a meter en total
total_elementos = int(input("Ingrese el número total de valores a introducir: "))

lista_original = []

#guardamos los números uno a uno en la lista
for i in range(1, total_elementos + 1):
    numero = float(input(f"Ingrese el número {i}: "))
    lista_original.append(numero)

# Sacamos el máximo, el mínimo y buscamos en qué posición estaban originalmente
val_max = max(lista_original)
pos_max = lista_original.index(val_max)

val_min = min(lista_original)
pos_min = lista_original.index(val_min)

#sacamos la media dividiendo la suma total entre la cantidad de elementos
promedio = sum(lista_original) / len(lista_original)

#hacemos una copia para ordenarla sin romper la lista original
lista_ordenada = sorted(lista_original)

#mostramos todo en pantalla
print("\nRESULTADOS:")
print(f"Lista original: {lista_original}")
print(f"Lista ordenada: {lista_ordenada}")
print(f"Promedio: {promedio}")
print(f"Valor máximo: {val_max}, en la posición {pos_max}")
print(f"Valor mínimo: {val_min}, en la posición {pos_min}")
```

8. Menú compra-venta

```python
#inicializamos las listas que usaremos 
compras = []
ventas = []
alquileres = []

#bucle infinito para que el menú se vuelva a mostrar hasta que elija salir
while True:
    print("\nMENU COMPRA-VENTA")
    print("1. Comprar")
    print("2. Vender")
    print("3. Alquilar")
    print("4. Mostrar listas de datos")
    print("5. Salir")
    
    opcion = input("Introduzca que desea hacer (1-5): ")

    if opcion == "1":
        #pido el producto a comprar y lo guardo en la lista de compras
        item = input("¿Qué quieres comprar?: ")
        compras.append(item)
        print(f"'{item}' guardado en la lista de compras.")
        
    elif opcion == "2":
        #pido el producto a vender y lo guardo en ventas
        item = input("¿Qué quieres vender?: ")
        ventas.append(item)
        print(f"'{item}' guardado en la lista de ventas.")
        
    elif opcion == "3":
        #pido el producto a alquilar y lo guardo en alquileres
        item = input("¿Qué quieres alquilar?: ")
        alquileres.append(item)
        print(f"'{item}' guardado en la lista de alquileres.")
        
    elif opcion == "4":
        #muestro el contenido de todas las listas que llevamos guardadas
        print("\n--- DATOS REGISTRADOS ---")
        print(f"Compras: {compras}")
        print(f"Ventas: {ventas}")
        print(f"Alquileres: {alquileres}")
        
    elif opcion == "5":
        #rompo el bucle para terminar el programa
        break
        
    else:
        #por si mete algo diferente a 1-5
        print("Opción no válida, intenta de nuevo.")
```

9. Programa de login con reintentos

```python
#bucle principal para reiniciar si falla 5 veces
while True:
    #guardamos la contraseña 
    password_correcta = input("Define la contraseña del sistema: ")
    
    #contador para controlar los 5 intentos
    intentos = 0
    login_exitoso = False
    
    #bucle para pedir la clave hasta 5 intentos
    while intentos < 5:
        intentos += 1
        password_ingresada = input(f"Introduce la contraseña (Intento {intentos}/5): ")
        
        if password_ingresada == password_correcta:
            print("Login correcto")
            login_exitoso = True
            break
        else:
            print(f"Login fallido. Intento {intentos}")
    
    #si logró loguearse, salimos del bucle principal
    if login_exitoso:
        break
    
    #si llegó aquí es porque consumió los 5 intentos sin acertar
    print("\nHas superado los 5 intentos fallidos. Reiniciando...\n")
```

10. Verificar aprobados y suspendidos con un diccionario

```python
#diccionario con los alumnos y sus notas
notas = {"Ana": 8, "Luis": 4, "Marta": 6, "Pedro": 3}

#recorremos el diccionario sacando la clave (alumno) y el valor (nota)
for alumno, nota in notas.items():
    if nota >= 5:
        print(f"{alumno}: {nota} - Aprobado")
    else:
        print(f"{alumno}: {nota} - Suspendido")
```
