1. Listas y Tuplas

```python
#Creacion de una lista y tupla
animes_lista = ["Naruto", "One Piece", "Attack on Titan"]
clubes_tupla = ("Barcelona", "Olympic", "Real Madrid")

#Imprimimos la lista y la tupla
print("Lista de animes:", animes_lista)
print("Tupla de clubes:", clubes_tupla)
print("--------------------------------")

#Mostramos el segundo elemento de la lista y el penúltimo elemento de la tupla
print("Segundo anime de la lista:", animes_lista[1])
print("Penúltimo club de la tupla:", clubes_tupla[-2])
print("--------------------------------")

#Si es que se puede, modificamos algun elemento de la lista y la tupla
animes_lista[0] = "Dragon Ball"
print("Lista de animes actualizada:", animes_lista)
#Como la tupla no se puede modificar, intentamos cambiar un elemento de la tupla y mostramos un mensaje de error
try:
    clubes_tupla[0] = "Juventus"
except TypeError:
    print("No se puede modificar un elemento de una tupla.")
print("--------------------------------")

#Mostramos el tamaño de la lista y la tupla
print("Tamaño de la lista de animes:", len(animes_lista))
print("Tamaño de la tupla de clubes:", len(clubes_tupla))
print("--------------------------------")

#Realizamos la busqueda de un elemento en la lista y la tupla a ver si devuelve true o false
anime_encontrado = "One Piece" in animes_lista
club_encontrado = "Olympic" in clubes_tupla
print("¿Está 'One Piece' en la lista de animes?", anime_encontrado)
print("¿Está 'Olympic' en la tupla de clubes?", club_encontrado)
print("--------------------------------")

#Si es que se puede, añadimos un elemento a la lista y la tupla, y mostramos un mensaje de error si no se puede
animes_lista.append("My Hero Academia")
try:
    clubes_tupla.append("Juventus")
except AttributeError:
    print("No se puede añadir un elemento a una tupla.")
print("Lista de animes actualizada:", animes_lista)
print("--------------------------------")

#Si es que se puede, eliminamosel contenido de la lista y la tupla, y mostramos un mensaje de error si no se puede
animes_lista.clear()
try:
    clubes_tupla.clear()
except AttributeError:
    print("No se puede eliminar el contenido de una tupla.")
```

2. Set y Diccionarios

```python
#Creamos el Set y Diccionario
cantantes_set = {"Alavaro Diaz", "Bad Bunny", "Nsqk"}
albumes_diccionario = {"Alavaro Diaz": "OMAKASE", "Bad Bunny": "DTMF", "Nsqk": "ATP"}
print("Set de cantantes:", cantantes_set)
print("Diccionario de álbumes:", albumes_diccionario)
print("--------------------------------")

#Mostramos el segundo elemento del set y el primer elemento del diccionario, hay q saber q los sets no podemos acceder a un elemento especidico del set y en el diccionarios podemos acceder mediante la clave
print("Segundo elemento del set de cantantes: No se puede acceder a un elemento específico de un set.")
print("Primer elemento del diccionario de álbumes:", albumes_diccionario["Alavaro Diaz"])
print("--------------------------------")

#Si es que se puede, modificamos algun elemento del set y el diccionario
cantantes_set.remove("Nsqk")
albumes_diccionario["Bad Bunny"] = "Un Verano Sin Ti"
print("Set de cantantes actualizado:", cantantes_set)
print("Diccionario de álbumes actualizado:", albumes_diccionario)
print("--------------------------------")

#Mostramos el tamaño del set y el diccionario
print("Tamaño del set de cantantes:", len(cantantes_set))
print("Tamaño del diccionario de álbumes:", len(albumes_diccionario))
print("--------------------------------")

#Realizamos la busqueda de un elemento en el set y el diccionario a ver si devuelve true o false
cantante_encontrado = "Bad Bunny" in cantantes_set
album_encontrado = "Nsqk" in albumes_diccionario
print("¿Está Bad Bunny en el set de cantantes?", cantante_encontrado)
print("¿Está Nsqk en el diccionario de álbumes?", album_encontrado)
print("--------------------------------")

#Si es que se puede, añadimos un elemento al set y el diccionario
cantantes_set.add("Rosalia")
albumes_diccionario["Rosalia"] = "Motomami"
print("Set de cantantes actualizado:", cantantes_set)
print("Diccionario de álbumes actualizado:", albumes_diccionario)
print("--------------------------------")

#Si es que se puede, eliminamos el contenido del set y el diccionario
cantantes_set.clear()
albumes_diccionario.clear()
print("Set de cantantes después de eliminar el contenido:", cantantes_set)
print("Diccionario de álbumes después de eliminar el contenido:", albumes_diccionario)
```

5. Manipulacion de listas de compras

```python
#Lista de productos
productos = ["Leche", "Pan", "Huevos", "Queso", "Mantequilla", "Jugo"]
print("Lista de productos:", productos)
print("--------------------------------")

#Mostramos el segundo elemento de la lista y el último elemento de la lista
print("Segundo producto de la lista:", productos[1])
print("Último producto de la lista:", productos[-1])
print("--------------------------------")

#Modificamos el tercer elemento de la lista y mostramos la lista actualizada
productos[2] = "Jamon"
print("Lista de productos actualizada:", productos)
print("--------------------------------")

#Intercambiamos el primer elemento de la lista con el último elemento de la lista
productos[0], productos[-1] = productos[-1], productos[0]
print("Lista de productos después del intercambio:", productos)
print("--------------------------------")

#Extraemos los 3 primeros elementos usando slicing
primeros_tres_productos = productos[:3]
print("Los 3 primeros productos de la lista son:", primeros_tres_productos)
print("--------------------------------")

#Extraemos los 3 últimos elementos usando slicing
ultimos_tres_productos = productos[-3:]
print("Los 3 últimos productos de la lista son:", ultimos_tres_productos)
print("--------------------------------")

#Extraemos los elemntos en posiciones pares usando slicing
productos_pares = productos[::2]
print("Los productos en posiciones pares de la lista son:", productos_pares)
```

6. Formateo de cadenas invertidas

```python
#Cadena invertida
cadena = "sajoR ordeP, 6002"

#Invertimos la cadena usando slicing
cadena_invertida = cadena[::-1]
print("Cadena original:", cadena)
print("Cadena invertida:", cadena_invertida)
print("--------------------------------")

#Separamos los elementos
partes = cadena_invertida.split(",")

nacimiento = partes[0]
nombre_apellido = partes[1]

#Mostramos los elementos separadosn y con el formato
print(f"Empleado {nombre_apellido} nacido en el año {nacimiento}")
```

7. Operaciones con conjuntos

```python
#Conjuntos
mamiferos = {"Gato", "León", "Caballo", "Ballena", "Tortuga"}
peces = {"Ballena", "Sardina"}

#Borramos la tortuga del conjuntos de mamiferos y añadimos un pez payaso al conjunto de peces
mamiferos.remove("Tortuga")
peces.add("Pez payaso")

print("Conjunto de mamíferos actualizado:", mamiferos)
print("Conjunto de peces actualizado:", peces)
print("--------------------------------")

#Mostramos los conjuntos repetidos de los dos juntos
ambos_conjuntos = mamiferos & peces
print("Elementos que hay en los dos conjuntos:", ambos_conjuntos)
```

8. Comparación de conjuntos de alumnos

```python
#Creamos dos conjuntos de alumnos
alumnos_python = {"Damian Rojas", "Marylen Chavez", "Leonardo Gutierres", "Lorenzo Quintana", "Luz Ojeda"}
alumnos_IA = {"Miguel Torres", "Daniel Arguello", "Sebastian Lopez", "Luz Ojeda", "Axel Martinez", "Mateo Rojas", "Damian Rojas"}

#Extraemos los alumnos que estan en los cursos y los que estan en ambos
en_ambos_cursos = alumnos_python & alumnos_IA
solo_python = alumnos_python - alumnos_IA
solo_IA = alumnos_IA - alumnos_python
print("Los alumnos que estan en los dos cursos: ", en_ambos_cursos)
print("Alumnos en el curso de Python: ", solo_python)
print("Alumnos en el curso de IA: ", solo_IA)
print("------------------------------------")

#Convertimos los dos conjuntos en una lista e imprimimos
alumnos_python_lista = list(solo_python)
alumnos_IA_lista = list(solo_IA)
en_ambos_cursos_lista = list(en_ambos_cursos)
print("Alumnos en ambos cursos (Lista):", en_ambos_cursos_lista)
print("Alumnos exclusivos de Python (Lista):", alumnos_python_lista)
print("Alumnos exclusivos de IA (Lista):", alumnos_IA_lista)
```

9. Depurando una lista de repetidos

```python
#Lista de elementos con valores duplicados
elementos_duplicados = ["manzana", "pera", "manzana", "naranja", "uva", "pera", "manzana"]
print("Lista original con duplicados:", elementos_duplicados)
print("---------------------------------------")

#Eliminamos los duplicados convirtiendo la lista a un set y lo guardamos 
conjunto_unicos = set(elementos_duplicados)
elementos_unicos = list(conjunto_unicos)

print("Lista depurada sin duplicados:", elementos_unicos)
```

10.  Gestión de tareas con un diccionario

```python
#Diccionario inicial con estructura predefinida
tareas = {
    "pendientes": [],
    "en_progreso": [],
    "completadas": []
}

tarea = input("Introduce una tarea: ")
estado = input("Introduce el estado (pendientes, en_progreso, completadas): ").strip().lower()
#Añadimos la tarea
tareas[estado].append(tarea)
print("--------------------------------------")

#Muestra del diccionario actualizado
print("Diccionario actualizado:")
print(tareas)
```
