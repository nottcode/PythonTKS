1. Función para calcular el área de un rectángulo

```python
# definimos la funcion que calcula el area recibiendo base y altura
def area_rectangulo(base, altura):
    # multiplicamos base por altura para sacar el area
    return base * altura

# le pedimos los datos al usuario por teclado
base_user = float(input("Introduce la base del rectángulo: "))
altura_user = float(input("Introduce la altura del rectángulo: "))

# llamamos a la funcion y guardamos el resultado
resultado = area_rectangulo(base_user, altura_user)

# imprimimos el resultado en pantalla
print(f"El área del rectángulo es: {resultado}")
```

2. Función para calcular el área de un círcul

```python
import math # importamos math para usar pi

# funcion para sacar el area del circulo dado su radio
def area_circulo(radio):
    # formula: pi * radio al cuadrado
    return math.pi * (radio ** 2)

# pedimos el radio por teclado
radio_user = float(input("Introduce el radio del círculo: "))

# ejecutamos la funcion y mostramos el resultado
area_result = area_circulo(radio_user)
print(f"El área del círculo es: {area_result}")
```

3. Función para encontrar el mayor de tres números

```python
# funcion auxiliar para pedir y leer un numero por teclado
def lee_numero(mensaje):
    return float(input(mensaje))

# funcion que compara los 3 numeros y devuelve el mas grande
def mayor(n1, n2, n3):
    if n1 >= n2 and n1 >= n3:
        return n1
    elif n2 >= n1 and n2 >= n3:
        return n2
    else:
        return n3

# reutilizamos lee_numero para pedir los 3 valores
num1 = lee_numero("Introduce el primer número: ")
num2 = lee_numero("Introduce el segundo número: ")
num3 = lee_numero("Introduce el tercer número: ")

# buscamos el mayor llamando a la funcion y mostramos
maximo = mayor(num1, num2, num3)
print(f"El mayor de los tres números es: {maximo}")
```

4. Cálculo del índice de masa corporal (IMC)

```python
# funcion que calcula el IMC y evalua el estado de la persona
def imc(peso, altura):
    # calculo de la formula del imc
    valor_imc = peso / (altura * altura)
    
    # evaluamos en que rango cae la persona
    if valor_imc < 18.50:
        clasificacion = "Bajo peso"
    elif 18.50 <= valor_imc < 25.00:
        clasificacion = "Normal"
    elif 25.00 <= valor_imc < 30.00:
        clasificacion = "Sobrepeso"
    else:
        clasificacion = "Obesidad"
        
    return valor_imc, clasificacion

# pedimos datos de peso y altura
peso_user = float(input("Introduce tu peso en kilos (ej. 70.5): "))
altura_user = float(input("Introduce tu altura en metros (ej. 1.75): "))

# llamamos a la funcion desestructurando los dos valores que devuelve
valor, estado = imc(peso_user, altura_user)

# imprimimos el imc redondeado a dos decimales y la clasificacion
print(f"Tu IMC es: {valor:.2f} | Clasificación: {estado}")
```

5. Login estructurado en funciones

```python
# funcion para pedir cualquier contraseña 
def pedir_contrasena(mensaje):
    return input(mensaje)

# funcion para gestionar el intento de validacion del login
def validar_login(password_correcta):
    intentos = 0
    # bucle para los 5 intentos maximos
    while intentos < 5:
        intentos += 1
        ingreso = pedir_contrasena(f"Introduce la contraseña (Intento {intentos}/5): ")
        
        if ingreso == password_correcta:
            print("Login correcto")
            return True
        else:
            print(f"Login fallido. Intento {intentos}")
            
    # si sale del bucle es que fallo los 5 intentos
    return False

# funcion principal que orquesta todo el flujo del login
def sistema_login():
    while True:
        # definir la clave
        clave_sistema = pedir_contrasena("Define la contraseña del sistema: ")
        
        # intentar loguearse
        exito = validar_login(clave_sistema)
        
        if exito:
            break
        else:
            print("\nHas alcanzado los 5 intentos fallidos. Reiniciando...\n")

# ejecutamos el sistema
sistema_login()
```

6. Función intermedio

```python
# funcion para sacar el medio de dos numeros
def intermedio(num1, num2):
    # la suma dividida entre 2 da el punto medio
    return (num1 + num2) / 2

# pedimos ambos numeros al usuario
n1 = float(input("Introduce el primer número: "))
n2 = float(input("Introduce el segundo número: "))

# calculamos el punto medio y mostramos
punto_medio = intermedio(n1, n2)
print(f"El punto intermedio entre {n1} y {n2} es: {punto_medio}")
```

7. Función recortar

```python
# funcion que limita un numero entre un minimo y un maximo
def recortar(numero, limite_inferior, limite_superior):
    if numero < limite_inferior:
        return limite_inferior
    elif numero > limite_superior:
        return limite_superior
    else:
        return numero

# pedimos los datos por teclado
num = float(input("Introduce el número a recortar: "))
lim_inf = float(input("Introduce el límite inferior: "))
lim_sup = float(input("Introduce el límite superior: "))

# ejecutamos la funcion recortar y mostramos el valor resultante
res = recortar(num, lim_inf, lim_sup)
print(f"El número recortado es: {res}")
```

8. Funcion separar()

```python
# funcion que separa una lista en dos listas ordenadas de pares e impares
def separar(lista_numeros):
    pares = []
    impares = []
    
    # recorremos cada numero de la lista original
    for n in lista_numeros:
        if n % 2 == 0:
            pares.append(n)
        else:
            impares.append(n)
            
    # ordenamos ambas listas de menor a mayor antes de devolverlas
    pares.sort()
    impares.sort()
    
    return pares, impares

# lista de ejemplo
numeros = [-12, 84, 13, 20, 33, 101, 9]

# llamamos a la funcion separar
lista_pares, lista_impares = separar(numeros)

# imprimimos el resultado esperado
print("Lista de pares ordenada:", lista_pares)
print("Lista de impares ordenada:", lista_impares)
```
